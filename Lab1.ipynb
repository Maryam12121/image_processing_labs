### Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

```python
import numpy as np

arr_2d = np.array([
    [5, 10, 15],
    [20, 25, 30],
    [35, 40, 45]
], np.uint8)

arr_2d
```

Retrieve the value 25 from the array using indexing.

```python
arr_2d[1, 1]
```

Extract the second row of the array.

```python
arr_2d[1]
```

Extract the third column of the array.

```python
arr_2d[:, 2]
```

Extract the bottom-right 2×2 sub-array from arr_2d.

```python
arr_2d[1:, 1:]
```

# Task #2: Loading and Visualizing Images


a.	Loading/reading image using OpenCv

```python
import cv2
img = cv2.imread('cameraman.tif')

#Load the image data into a NumPy array
import matplotlib.pyplot as plt
plt.imshow(img)
plt.show()
```

b.	Loading/reading image using PIL

```python
from PIL import Image
img2 = Image.open('lena_gray_256.jpg') #load image into a PIL Image object

import matplotlib.pyplot as plt
import matplotlib.cm as cm
plt.imshow(img2, cmap = cm.Greys_r)
plt.show()
```
